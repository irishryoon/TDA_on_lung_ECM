```julia
using CSV
using DataFrames
using Plots

gr()

# parameters for point cloud plotting 
p_markersize = 5
p_markerstrokewidth = 0.1
p_imagesize = (300, 300)
```

# Plot ECM point cloud, cancer cells, and leukocytes

```julia
ROI = 2

# load the ECM point cloud
ECM_points = CSV.read("../data/pointcloud/ECM/ROI" * string(ROI) * ".csv", DataFrame);
celltype1 = CSV.read("../data/pointcloud/cancer/ROI" * string(ROI) * ".csv", DataFrame)
celltype2 = CSV.read("../data/pointcloud/leukocytes/ROI" * string(ROI) * ".csv", DataFrame)


p1 = scatter(ECM_points.x, ECM_points.y, 
                yflip = :true, 
                c = "#259ea1", 
                label = "", 
                frame = :box, 
                #ticks = [], 
                markersize = 5, 
                aspect = :equal,
                size = p_imagesize,
                background_color=:transparent, 
                foreground_color=:black)

# plot the two cell types
p2 = scatter(celltype1[:,1], celltype1[:,2],
                markersize = p_markersize,
                    yflip = true,
                    label = "",
                    markerstrokewidth = p_markerstrokewidth,
                    frame = :box,
                    #ticks = [],
                    c =  "#592693", # purple
                    aspect = :equal, size = p_imagesize,  
                    background_color=:transparent, foreground_color=:black
                    )
p3 = scatter(celltype2[:,1], celltype2[:,2],
            markersize = p_markersize,
                yflip = true,
                label = "",
                markerstrokewidth = p_markerstrokewidth,
                frame = :box,
                #ticks = [],
                c = "#FAAB36", # orange
                 aspect = :equal, size = p_imagesize,  
                background_color=:transparent, foreground_color=:black
                )


plot(p1, p2, p3, layout = grid(1,3), size = (900, 300))
```
